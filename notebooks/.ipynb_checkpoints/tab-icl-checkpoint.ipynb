## Imports and version check

```python
import numpy as np
import torch
import sklearn
import tabicl
from tabicl import TabICLClassifier

print("tabicl      ", tabicl.__version__)
print("torch       ", torch.__version__)
print("scikit-learn", sklearn.__version__)
print("device      ", "cuda" if torch.cuda.is_available() else "cpu")
```

## 3. A dataset defined inline

A tiny, transparent binary-classification table generated in the notebook (seeded, so
it is reproducible). Two informative features drive the label; we hold out a few rows
as the test set that TabICL will predict.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

# in-context "training" examples + held-out rows to predict
X, y = make_classification(
    n_samples=200,
    n_features=5,
    n_informative=3,
    n_redundant=0,
    n_classes=2,
    random_state=42,
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=8, random_state=42, stratify=y
)

print("context (train) rows:", X_train.shape, "  predict (test) rows:", X_test.shape)
print("first context row    :", np.round(X_train[0], 3), "-> label", y_train[0])
```

## 4. Fit and predict

`fit()` just memorizes the context (no gradient training) and downloads the checkpoint
on first use; `predict()` runs the in-context forward pass.

```python
clf = TabICLClassifier()
clf.fit(X_train, y_train)  # downloads the checkpoint from Hugging Face on first run

y_pred = clf.predict(X_test)
y_proba = clf.predict_proba(X_test)

for i, (pred, true) in enumerate(zip(y_pred, y_test)):
    mark = "ok " if pred == true else "MISS"
    print(f"row {i}: pred={pred} true={true} p={y_proba[i].round(3)}  {mark}")

print("\naccuracy on held-out rows:", (y_pred == y_test).mean())
```
