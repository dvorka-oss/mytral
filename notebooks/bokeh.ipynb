```python
import bokeh
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

print("bokeh", bokeh.__version__)

# render plots inline instead of opening a browser tab
output_notebook()
```

## Example 1: bar chart

A categorical bar chart - pass the categories as the figure's `x_range`, then draw
vertical bars with `vbar`.

```python
fruits = ["Apples", "Pears", "Nectarines", "Plums", "Grapes", "Strawberries"]
counts = [5, 3, 4, 2, 4, 6]

p = figure(
    x_range=fruits,
    height=350,
    title="Fruit counts",
    toolbar_location=None,
    tools="",
)
p.vbar(x=fruits, top=counts, width=0.9, color="#1f77b4")

p.xgrid.grid_line_color = None
p.y_range.start = 0
p.yaxis.axis_label = "count"

show(p)
```

## Example 2: scatter plot

A scatter of two continuous variables (seeded random data), drawn with `scatter`.

```python
import numpy as np

rng = np.random.default_rng(42)
x = rng.normal(size=200)
y = 0.7 * x + rng.normal(scale=0.5, size=200)

p = figure(
    height=350,
    width=500,
    title="Scatter: y ~ 0.7x + noise",
    x_axis_label="x",
    y_axis_label="y",
)
p.scatter(x, y, size=8, alpha=0.5, color="navy")

show(p)
```
